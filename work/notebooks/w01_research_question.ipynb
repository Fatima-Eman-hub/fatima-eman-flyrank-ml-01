## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

**Search question:** Given observable search and content signals, which pages should be reviewed first for refresh, expansion, or monitoring?

**Unit of analysis:** one content page (`content_hash_id`), evaluated using its observed signals over a defined window (starter data uses a 90-day window).

**Output:** a ranked list of pages, each with a numeric opportunity/risk score and a reason code explaining why it was flagged (e.g. `stale_visible_page`, `declining_with_demand`).

**Action someone could take:** a content strategist or writer picks pages from the top of the ranked queue and reviews them first — refreshing content, updating structure, or expanding thin sections — instead of guessing which pages need attention.

**Cost of a wrong recommendation:** if the system flags a page that isn't actually a priority, the team spends limited review time/budget on a page that didn't need it, while a genuinely declining page waits longer. This is a resource-misallocation cost, not a safety-critical one — but it's real, since review capacity is limited and every wrong pick displaces a page that did need attention.

**Why data/ML can help:** a simple hand-written rule (e.g. "stale AND visible") can only combine a couple of signals crudely. The starter results show a learned model finds a combination of signals a fixed rule misses, and — critically — the result is still readable and explainable (Notebook 02 showed a full decision tree in a few lines of if/else), so this isn't a black box replacing judgment; it's a way to rank more pages more consistently than a person manually scanning a spreadsheet could.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import os, sys, subprocess

IN_COLAB = "google.colab" in sys.modules
REPO_URL = "https://github.com/flyrank-bih/flyrank-ml-internship-starter"
REPO_DIR = "flyrank-ml-internship-starter"

if IN_COLAB:
    if not os.path.isdir(REPO_DIR):
        subprocess.run(["git", "clone", "--depth", "1", REPO_URL, REPO_DIR], check=True)
    os.chdir(REPO_DIR)
elif os.path.basename(os.getcwd()) in ("notebooks", "work"):
    os.chdir("..")

import pandas as pd
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

# Number 1: how many pages qualify as "declining with demand" — a candidate reason code for this lane
declining_with_demand = ((df["trend_direction"] == "down") & (df["impressions_90d"] >= 100)).sum()
print(f"Pages matching 'declining_with_demand': {declining_with_demand} of {len(df)} "
      f"({declining_with_demand/len(df):.1%})")

# Number 2: how many pages are stale AND still visible — the other candidate reason code
stale_visible = ((df["days_since_last_update"] >= 180) & (df["impressions_90d"] >= 500)).sum()
print(f"Pages matching 'stale_visible_page': {stale_visible} of {len(df)} "
      f"({stale_visible/len(df):.1%})")

# Number 3: documented starter pipeline result (baseline vs learned model), from the lane guide
print("\nDocumented starter pipeline result (docs/ml-intern-dataset-and-lane-guide.md):")
print("Baseline rule   Precision@50 = 0.240  (~12 of top 50 correct)")
print("Random forest   Precision@50 = 0.740  (~37 of top 50 correct)")
```

**What these numbers show:** a meaningful slice of the starter dataset already fits the two simplest reason codes for this lane — meaning the review queue would have real candidates to rank from day one, not an empty or trivial list. And on the documented starter pipeline result, the baseline rule scores well below the random forest at Precision@50 — evidence that learned ranking beats a fixed rule on this exact task shape, which is why this lane looks worth pursuing for the next 7 weeks.
